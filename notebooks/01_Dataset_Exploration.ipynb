```python
# Step 1: Checking Missing Values

# Missing values are one of the most common problems in real-world datasets.
# Before training a Machine Learning model, we must identify columns
# containing missing values.

import pandas as pd

missing_data = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_data
```

```python
# Step 2: Visualizing Missing Values

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

# Visualize missing values in the dataset
sns.heatmap(df.isnull(), cbar=False, yticklabels=False, cmap="viridis")

plt.title("Missing Values in Crop Yield Dataset")
plt.xlabel("Features")
plt.ylabel("Records")
plt.show()
```

```python
# Step: Checking Duplicate Rows

# Duplicate rows can affect the performance of a Machine Learning model.
# Therefore, we check for duplicate records before preprocessing the dataset.

duplicate_rows = df[df.duplicated()]

print(f"Total duplicate rows detected: {len(duplicate_rows)}")

print(duplicate_rows)

print("-" * 40)
```

```python
# Display rows where missing values are present

missing_columns = [
    'Annual_Rainfall',
    'Fertilizer',
    'Pesticide',
    'Yield',
    'Avg_Temperature'
]

df[df[missing_columns].isnull().any(axis=1)]
```

### Class Distribution

Class distribution is **not applicable** to this project because this is a regression problem. The target variable `Yield` is a continuous numerical variable rather than a set of discrete classes.

```python
## Target Variable Distribution

# The target variable is Yield.
# Since this is a regression problem, we examine the distribution
# of Yield values to understand the range and spread of crop yield.

df["Yield"].describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.histplot(df["Yield"], kde=True)

plt.title("Distribution of Crop Yield")
plt.xlabel("Yield")
plt.ylabel("Frequency")

plt.show()
```

```python
# Percentage Distribution of Yield

yield_ranges = pd.cut(
    df["Yield"],
    bins=3,
    labels=["Low", "Medium", "High"]
)

(yield_ranges.value_counts(normalize=True) * 100).round(2)
```
