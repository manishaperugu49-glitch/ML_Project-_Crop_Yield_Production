```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

print("Libraries imported successfully!")
```

```python
df = pd.read_csv(r"C:\Crop Yield Prediction\data\Crop Yield Data.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
missing_data = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_data
```

```python
# Handle missing values

numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_columns = df.select_dtypes(
    include=["object"]
).columns

# Fill missing numerical values with median
for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())

# Fill missing categorical values with most frequent value
for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

print("Missing values treated successfully!")
print("\nMissing values after treatment:")
print(df.isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

```python
# Check for duplicate records

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("Duplicate records removed successfully!")
else:
    print("No duplicate records found.")

print("Dataset shape after duplicate removal:", df.shape)
```

```python
df = df.drop_duplicates().reset_index(drop=True)

print("Duplicates removed successfully!")
print("New shape:", df.shape)
```

```python
X = df.drop("Yield", axis=1)
y = df["Yield"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
print("Input Features:")
print(X.columns.tolist())

print("\nTarget Variable:")
print("Yield")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

```python
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
# Numerical preprocessing

numerical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

print("Numerical preprocessing pipeline created!")
```

```python
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

print("Categorical preprocessing pipeline created!")
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline, numerical_features),
        ("cat", categorical_pipeline, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
preprocessor.fit(X_train)

print("Preprocessor fitted using training data only!")
```

```python
X_train_processed = preprocessor.transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Training data processed successfully!")
print("Testing data processed successfully!")

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

## Train-Test Split

The dataset is divided into training and testing sets using an 80:20 ratio.

The training set is used to learn the preprocessing parameters and train the machine learning models, while the testing set is kept separate for final model evaluation.

A fixed random state of 42 is used to ensure reproducibility.

```python
from sklearn.model_selection import train_test_split

X = df.drop("Yield", axis=1)
y = df["Yield"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

```python
numerical_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

numerical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

print("Numerical preprocessing pipeline created successfully!")
print("Steps:")
print("1. Median imputation")
print("2. StandardScaler")
```

```python
from sklearn.preprocessing import OneHotEncoder

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

print("Categorical preprocessing pipeline created successfully!")
print("Steps:")
print("1. Most-frequent-value imputation")
print("2. One-hot encoding")
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline, numerical_features),
        ("cat", categorical_pipeline, categorical_features)
    ]
)

print("Combined preprocessing pipeline created successfully!")
```

```python
# Fit preprocessing only on training data

preprocessor.fit(X_train)

print("Preprocessor fitted using training data only.")
print("Test data was not used to learn imputation or scaling parameters.")
```

```python
X_train_processed = preprocessor.transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Training data processed successfully!")
print("Testing data processed successfully!")

print("\nProcessed Training Shape:", X_train_processed.shape)
print("Processed Testing Shape:", X_test_processed.shape)
```

## Preprocessing Conclusion

The Crop Yield Dataset was successfully prepared for machine learning.

Duplicate records were identified and removed where applicable. Missing numerical values were handled using median imputation, while missing categorical values were handled using the most frequent value.

Categorical variables were converted into numerical form using One-Hot Encoding. Numerical variables were standardized using StandardScaler.

# Feature Selection

Feature selection is performed to identify the variables that are useful for predicting crop yield.

The selection process considers:
- Correlation with the target variable
- Agricultural relevance
- Data leakage
- Availability of the feature at prediction time

Features that contain target leakage will not be used for model training.

## Feature Selection Reasoning

The features are selected based on their relevance to crop yield prediction and their availability during prediction.

### Features considered

- Crop: identifies the crop type.
- Crop_Year: represents the agricultural year.
- Season: represents the growing season.
- State: represents geographical variation.
- Area: represents cultivated area.
- Production: checked separately for possible target leakage.
- Annual_Rainfall: represents rainfall conditions.
- Fertilizer: represents fertilizer usage.
- Pesticide: represents pesticide usage.
- Avg_Temperature: represents average temperature.
- Max_Temperature: represents maximum temperature.
- Min_Temperature: represents minimum temperature.

Production is specifically checked for target leakage because it may have a mathematical relationship with Yield and Area.
