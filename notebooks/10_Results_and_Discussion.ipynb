# Results and Discussion

This notebook presents the results obtained from the Random Forest,
XGBoost, and SVR models for Crop Yield Prediction.

The models are analyzed using MAE, MSE, RMSE, and R² score.
The strengths, limitations, prediction errors, and overall findings
are discussed based on the experimental results.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
```

## Loading Model Comparison Results

The model comparison results from the previous notebook are loaded to
analyze the performance of Random Forest, XGBoost, and SVR using MAE,
MSE, RMSE, and R² Score.

```python
comparison_path = r"C:\Crop Yield Prediction\outputs\model_comparison.csv"

comparison_df = pd.read_csv(comparison_path)

print("Model comparison results:")
display(comparison_df)
```

## Performance Comparison

The three regression models are compared using four evaluation metrics:

- **MAE (Mean Absolute Error):** Measures the average absolute difference
  between actual and predicted yield values.
- **MSE (Mean Squared Error):** Measures the average squared prediction error.
- **RMSE (Root Mean Squared Error):** Represents the prediction error in the
  same unit as the target variable.
- **R² Score:** Measures how well the model explains the variation in crop yield.

For MAE, MSE, and RMSE, lower values represent smaller prediction errors.
For R² Score, a higher value represents a greater proportion of variation
explained by the model.

```python
display(
    comparison_df.style.format({
        "MAE": "{:.4f}",
        "MSE": "{:.4f}",
        "RMSE": "{:.4f}",
        "R2": "{:.4f}"
    })
)
```

## MAE Comparison

The following graph compares the Mean Absolute Error of the three models.
A lower MAE indicates smaller average prediction errors.

```python
plt.figure(figsize=(8, 5))

plt.bar(
    comparison_df["Model"],
    comparison_df["MAE"]
)

plt.xlabel("Model")
plt.ylabel("MAE")
plt.title("MAE Comparison of Regression Models")

plt.tight_layout()
plt.show()
```

## RMSE Comparison

The following graph compares the Root Mean Squared Error of the models.
Lower RMSE values indicate smaller prediction errors.

```python
plt.figure(figsize=(8, 5))

plt.bar(
    comparison_df["Model"],
    comparison_df["RMSE"]
)

plt.xlabel("Model")
plt.ylabel("RMSE")
plt.title("RMSE Comparison of Regression Models")

plt.tight_layout()
plt.show()
```

## R² Score Comparison

The following graph compares the R² Score of the three regression models.
A higher R² Score indicates that the model explains a larger proportion
of the variation in crop yield.

```python
plt.figure(figsize=(8, 5))

plt.bar(
    comparison_df["Model"],
    comparison_df["R2"]
)

plt.xlabel("Model")
plt.ylabel("R² Score")
plt.title("R² Score Comparison of Regression Models")

plt.tight_layout()
plt.show()
```
