# Final Prediction / Demonstration

This notebook demonstrates crop yield prediction using the final selected
machine learning model.

The trained model and preprocessing pipeline are loaded from the models
folder. A sample unseen input is provided to the model, and the predicted
crop yield is displayed as the output.

No model training is performed in this notebook.

## Final Model

The final selected model from the model evaluation and comparison process
is used for prediction.

The saved model and its corresponding preprocessing pipeline are loaded
from the project directory.

```python
import pandas as pd
import joblib
```

```python
comparison_path = r"C:\Crop Yield Prediction\outputs\model_comparison.csv"

comparison_df = pd.read_csv(comparison_path)

display(comparison_df)
```

## Selecting the Final Model

The final model is selected based on the model comparison results.

For regression evaluation, MAE, MSE, and RMSE represent prediction error,
where lower values indicate smaller errors. R² Score represents the amount
of variation explained by the model, where a higher value indicates better
explanatory performance.

The model selected here will be used for the final prediction demonstration.

```python
best_model_name = comparison_df.loc[
    comparison_df["RMSE"].idxmin(),
    "Model"
]

print("Final selected model:", best_model_name)
```

```python
model_paths = {
    "Random Forest": r"C:\Crop Yield Prediction\models\random_forest_model.pkl",
    "XGBoost": r"C:\Crop Yield Prediction\models\xgboost_model.pkl",
    "SVR": r"C:\Crop Yield Prediction\models\svr_model.pkl"
}

if best_model_name == "XGBoost":
    final_model = joblib.load(
        r"C:\Crop Yield Prediction\models\xgboost_model.pkl"
    )
    final_preprocessor = joblib.load(
        r"C:\Crop Yield Prediction\models\tuned_preprocessor.pkl"
    )

elif best_model_name == "Random Forest":
    final_model = joblib.load(
        r"C:\Crop Yield Prediction\models\random_forest_model.pkl"
    )
    final_preprocessor = joblib.load(
        r"C:\Crop Yield Prediction\models\random_forest_preprocessor.pkl"
    )

elif best_model_name == "SVR":
    final_model = joblib.load(
        r"C:\Crop Yield Prediction\models\svr_model.pkl"
    )
    final_preprocessor = joblib.load(
        r"C:\Crop Yield Prediction\models\svr_preprocessor.pkl"
    )

print("Final selected model:", best_model_name)
print("Final model loaded successfully.")
print("Preprocessing pipeline loaded successfully.")
```

## Loading the Final Model

The saved trained model corresponding to the selected model is loaded
from the models directory.

The preprocessing pipeline used during model training is also loaded so
that the new input is transformed in the same way as the training data.

## Sample Unseen Input

A new sample is created to demonstrate prediction.

The input contains crop, year, season, state, area, rainfall, fertilizer,
pesticide, and temperature-related features.

The target variable `Yield` is not included because it is the value that
the model must predict.

```python
new_data = pd.DataFrame({
    "Crop": ["Rice"],
    "Crop_Year": [2025],
    "Season": ["Kharif"],
    "State": ["Telangana"],
    "Area": [500],
    "Annual_Rainfall": [1000],
    "Fertilizer": [150],
    "Pesticide": [10],
    "Avg_Temperature": [27],
    "Max_Temperature": [34],
    "Min_Temperature": [21]
})

display(new_data)
```

## Crop Yield Prediction

The new input is passed through the saved preprocessing pipeline and then
to the final trained model.

The resulting value represents the predicted crop yield for the given
unseen input.

```python
new_data_processed = final_preprocessor.transform(new_data)

predicted_yield = final_model.predict(new_data_processed)

print("Predicted Crop Yield:", predicted_yield[0])
```

```python
prediction_result = new_data.copy()

prediction_result["Predicted_Yield"] = predicted_yield

display(prediction_result)
```

The final trained model successfully generated a crop yield prediction
for a new input sample that was not used during model training.

The demonstration shows how the completed machine learning pipeline can
accept crop and environmental information and produce an estimated crop
yield.

This final prediction demonstrates the practical application of the
